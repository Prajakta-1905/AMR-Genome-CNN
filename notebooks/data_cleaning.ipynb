## Step 6 — Check for phenotype conflicts

```python
phenotype_by_genome = (
    df.groupby("Genome ID")["Resistant Phenotype"]
      .apply(lambda x: set(x.dropna()))
)

conflicting_genomes = phenotype_by_genome[
    phenotype_by_genome.apply(
        lambda x: "Resistant" in x and "Susceptible" in x
    )
]

print("Genomes with Resistant/Susceptible conflicts:",
      len(conflicting_genomes))
```

## Step 7 — Investigate duplicate Genome IDs
🧬 Cell 7.1 — Summarize every duplicated genome

```python
duplicate_summary = (
    df[df["Genome ID"].duplicated(keep=False)]
    .groupby("Genome ID")
    .agg(
        records=("Genome ID", "size"),
        phenotypes=("Resistant Phenotype", lambda x: set(x.dropna())),
        measurements=("Measurement", lambda x: list(x.dropna())),
        measurement_units=("Measurement Unit", lambda x: set(x.dropna())),
    )
    .sort_index()
)

duplicate_summary.head(20)
```

## Step 7.2 — See how many records each duplicated genome has

```python
print("Distribution of records per duplicated Genome ID:")
print(duplicate_summary["records"].value_counts().sort_index())
```

## Step 7.3 — Check phenotype patterns

```python
phenotype_patterns = (
    df.groupby("Genome ID")["Resistant Phenotype"]
    .agg(lambda x: tuple(sorted(set(x.dropna()))))
)

print("Phenotype patterns across all genomes:")
print(phenotype_patterns.value_counts())
```

## Step 7.4 — Find genomes with no phenotype

```python
missing_phenotype_genomes = phenotype_patterns[
    phenotype_patterns == ()
]

print("Genomes with no phenotype:", len(missing_phenotype_genomes))
```

## Step 8 — Define the binary classification cohort

For the CNN, the prediction task is binary classification of ciprofloxacin resistance.

Therefore:
- Resistant → retained as class 1
- Susceptible → retained as class 0
- Intermediate → excluded because it does not belong to either binary class
- Missing phenotype → excluded because no training label is available

```python
print("Usable binary genomes:")

binary_genomes = phenotype_patterns[
    phenotype_patterns.isin([
        ("Resistant",),
        ("Susceptible",)
    ])
]

print("Total:", len(binary_genomes))
print("\nClass distribution:")
print(binary_genomes.value_counts())
```

## Step 9 — Create one record per genome

Each genome must appear exactly once in the final binary dataset.

The phenotype is encoded as:
- 0 = Susceptible
- 1 = Resistant

Intermediate and missing-phenotype genomes are excluded from the binary classification cohort.

```python
# Create one row per genome using the verified phenotype patterns

clean_genomes = pd.DataFrame({
    "Genome ID": binary_genomes.index,
    "Phenotype": binary_genomes.values
})

# Convert phenotype tuples to binary labels

clean_genomes["Label"] = clean_genomes["Phenotype"].map({
    ("Susceptible",): 0,
    ("Resistant",): 1
})

clean_genomes = clean_genomes.sort_values("Genome ID").reset_index(drop=True)

print("Clean genome-level dataset:")
print("Total genomes:", len(clean_genomes))

print("\nLabel distribution:")
print(clean_genomes["Label"].value_counts().sort_index())
```

```python
print("Unique Genome IDs:", clean_genomes["Genome ID"].nunique())
print("Duplicate Genome IDs:", clean_genomes["Genome ID"].duplicated().sum())

print("\nFirst 10 genomes:")
print(clean_genomes.head(10))
```

## Step 10 — Save the clean genome-level dataset

The cleaned genome-level dataset is saved separately from the raw data so that the original dataset remains unchanged and the preprocessing step is reproducible.

```python
# Save the cleaned genome-level dataset

output_path = "../data/processed/ciprofloxacin_genome_level.csv"

clean_genomes.to_csv(output_path, index=False)

print("Clean dataset saved to:", output_path)
print("Rows saved:", len(clean_genomes))
```
